```python
#m=7
#n=6

def print_solution(S, X, i, j):
    #print("---------", i, j)
    #print(S[i][j])
    if i==0 or j==0:
        return
    if S[i][j] == "V":
        print ("-------*--", X[i])
        print_solution(S, X, i-1, j-1)
    elif S[i][j] == "U":
        print_solution(S, X, i-1, j)
    else:
        print_solution(S, X, i, j-1)

def LCS_bottom_up(X,Y,m,n):
    C = [ [0 for i in range(n+1)] for j in range(m+1) ]  
    S = [ [0 for i in range(n+1)] for j in range(m+1) ]  
    #print(C)
    XFIRST = X
    YFIRST = Y
    X = " " + X
    Y = " " + Y
    C[0][0] = 0
    S[0][0] = 0
    
    for i in range(1, m+1):
        C[i][0]=0

    for j in range(0, n+1):
        C[0][j]=0	

    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i] == Y[j]:
                C[i][j] = C[i-1][j-1] + 1
                S[i][j] = "V"
            elif C[i-1][j] >= C[i][j-1]:
                C[i][j] = C[i-1][j]
                S[i][j] = "U"
            else:
                C[i][j] = C[i][j-1] 
                S[i][j] = "L"
    #print(C)
    print_table(C, X, YFIRST)  
    print("------------------------------------")
    print_table(S, X, YFIRST)
    print_solution(S, X, m, n)

    return C[m][n]

def print_table(C, P, Q):
    print("          ", end="")
    for c in Q:
        print(f"{c:3}", end=" ")
    print()

    for i in range(len(C)):
        print(f"{P[i]:>3}", end=" ")
        for j in range(len(C[0])):
            print(f"{C[i][j]:3}", end=" ")
        print()

X="BDCABA"
Y="ABCBDAB"

LCS_bottom_up(X, Y, len(X), len(Y) )
```
