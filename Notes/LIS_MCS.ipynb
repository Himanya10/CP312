```python
A = [5, 2, 8, 6, 3, 6, 9, 7]
n=8
L = [1 for i in range(n)]
S = [-1 for i in range(n)]

def print_solution(n):
    #find the maximum index
    k = 0
    for i in range(0, n):
        if L[i] > L[k]:
            k = i
    i=k
    while i !=-1:        
        print(A[i])
        i = S[i] 

def LIS(n):
    #Initialization
    for i in range(1, n):        
        for j in range(0, i):
            if A[i] > A[j] and L[i] < L[j]+1:            
                L[i] = L[j]+1
                S[i] = j

    return max (L)

print ( LIS(8) )
print(L)
print(S)
print_solution(n)

```

```python
#Maximum Contiguous Subarray: First Approach
A=[1, 2, 3, -7, 3, 1, 1, -4, 6, -3, 1]
#A=[1, -7, 4, 0, 2, -1, 3, 0]
n = len(A)
ME = [1 for i in range(n)]
ME[0]=A[0]

def MS(n):
    for i in range(1,n):
        ME[i] = max ( A[i], A[i] + ME[i-1] )        
    
    return max( ME )
print(MS(n))
```

```python
#Maximum Contiguous Subarray: Second Approach
A=[1, 2, 3, -7, 3, 1, 1, -4, 6, -3, 1]
#A=[1, -7, 4, 0, 2, -1, 3, 0]
n = len(A)
M = [1 for i in range(n)]
ME = [1 for i in range(n)]
S = [1 for i in range(n)]
M[0] = max(0, A[0])
ME[0]=A[0]

def MS(n):
    for i in range(1,n):
        ME[i] = max ( A[i], A[i] + ME[i-1] )
        M[i] = max( ME[i], M[i-1] )
    return max(M[n-1], ME[n-1])
    
print(MS(n))
```
