```python
def Edit_Distance(X,Y,m,n):
    M = [ [0 for i in range(n+1)] for j in range(m+1) ]  
    #print(C)
    XFIRST = X
    YFIRST = Y
    X = " " + X
    Y = " " + Y
    M[0][0]=0
    for i in range(1, m+1):
        M[i][0]=i

    for j in range(0, n+1):
        M[0][j]=j	

    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i] == Y[j]:
                M[i][j] = M[i-1][j-1] 
            else:               
                M[i][j] = min ( M[i-1][j], M[i][j-1], M[i-1][j-1] ) + 1
    #print(C)
    print_table(M, X, YFIRST)    
    return M[m][n]

def print_table(M, P, Q):
    print("          ", end="")
    for c in Q:
        print(f"{c:3}", end=" ")
    print()

    for i in range(len(M)):
        print(f"{P[i]:3}", end=" ")
        for j in range(len(M[0])):
            print(f"{M[i][j]:3}", end=" ")
        print()

#X="SNOWY"
#Y="SUNNY"

X = "LILY"
Y = "LILLY"

Edit_Distance(X, Y, len(X), len(Y) )
```
